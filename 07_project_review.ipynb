	检查关键文件并记录环境

```python
from pathlib import Path
import sys

project_dir = Path.cwd()

required_files = [
    "BUCHWALD_HARTWIG_PLAN.md",
    "data/raw/Dreher_and_Doyle_input_data.xlsx",
    "data/splits/split_manifest.csv",
    "01_data_exploration.ipynb",
    "02_feature_engineering.ipynb",
    "03_model_training.ipynb",
    "04_model_evaluation.ipynb",
    "05_shap_analysis.ipynb",
    "06_improvement_experiments.ipynb",
    "reports/morgan_random_forest/random_model.joblib",
    "reports/morgan_random_forest/additive_model.joblib",
    "reports/evaluation/metrics.csv",
    "reports/shap/explanation_config.json",
    "reports/improvement/morgan_radius_comparison.csv",
]

print("Project directory:", project_dir)
print("Python executable:", sys.executable)

missing_files = []

for relative_path in required_files:
    exists = (project_dir / relative_path).is_file()
    print(f"{'OK' if exists else 'MISSING'} | {relative_path}")
    if not exists:
        missing_files.append(relative_path)

print("\nMissing file count:", len(missing_files))
```

```python
import subprocess

environment = subprocess.run(
    [sys.executable, "-m", "pip", "freeze"],
    check=True,
    capture_output=True,
    text=True,
)

(project_dir / "requirements.txt").write_text(
    environment.stdout,
    encoding="utf-8",
)

(project_dir / "python_version.txt").write_text(
    sys.version + "\n",
    encoding="utf-8",
)

print("Saved requirements.txt and python_version.txt")
```
